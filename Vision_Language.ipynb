##  Images
*  pixel digitization. (converting color to numers )
* Black and White (Binary): The simplest grid. A 0 represents pure black, and a 1 represents pure white.
* Grayscale: The grid uses numbers typically ranging from 0 to 255. 0 is black, 255 is white, and everything in between (1 to 254) represents different shades of grey.
* Color (RGB): Most modern images use the RGB model. Instead of just one grid, the computer stacks
   * three layers of grids together:
        * One grid for Red
        * One grid for Green
        * One grid for Blue
* pixels with Red, Green, and Blue (RGB) values.

## Videos

*  a sequence of image grids played rapidly over time, combined with a synchronized stream of audio data.
* Frames: **Every individual picture** in the video is **a frame**. Each frame is a standard color RGB grid of numbers .
* **FPS (Frames Per Second)**: This is the **speed** at which the computer changes the grids. Standard videos usually play at 24, 30, or 60 FPS. At 60 FPS, the computer is processing 60 entirely separate grids of numbers every single second.
